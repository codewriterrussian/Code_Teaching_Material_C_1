## 使用方式

1. 用 **Jupyter Notebook／JupyterLab／VS Code 的 Python 3 kernel** 開啟 `.ipynb`，本機須安裝 `cc`、`clang` 或 `gcc`。**作答語言是 C11，不是 Python。**
2. 首先執行下方「自動測試器」的 Python 程式格。
3. 依序閱讀題目、範例與限制；在 `%%writefile` 的格子撰寫完整 C 程式（包括 `main`）。
4. **每次修改後，先重新執行 `%%writefile` 程式格**，再執行下一格的 `run_public(題號)`。
5. 每題測試包含 **1 組公開範例＋5 組額外測資**。額外測資不會直接顯示預期答案；本地測試不是防作弊的正式評測系統。

````python
# 共用自動測試器｜先執行這一格（請勿修改）
from pathlib import Path
import subprocess, tempfile, shutil, json, base64, csv

_ANSWER_DIR=Path('c_loop_array_50_answers')
_ANSWER_DIR.mkdir(parents=True,exist_ok=True)
_COMPILER=shutil.which('cc') or shutil.which('gcc') or shutil.which('clang')
if _COMPILER is None:
    print('找不到 C 編譯器。請先安裝 GCC 或 Clang，重開 kernel 後再執行。')
else:
    print('C 編譯器：', _COMPILER)

_TEST_CASES=json.loads(base64.b64decode('eyIxIjpbeyJpbnB1dCI6IjVcbiIsIm91dHB1dCI6IjEgMiAzIDQgNVxuIn0seyJpbnB1dCI6IjFcbiIsIm91dHB1dCI6IjFcbiJ9LHsiaW5wdXQiOiIyXG4iLCJvdXRwdXQiOiIxIDJcbiJ9LHsiaW5wdXQiOiIzXG4iLCJvdXRwdXQiOiIxIDIgM1xuIn0seyJpbnB1dCI6IjEwXG4iLCJvdXRwdXQiOiIxIDIgMyA0IDUgNiA3IDggOSAxMFxuIn0seyJpbnB1dCI6Ijk5XG4iLCJvdXRwdXQiOiIxIDIgMyA0IDUgNiA3IDggOSAxMCAxMSAxMiAxMyAxNCAxNSAxNiAxNyAxOCAxOSAyMCAyMSAyMiAyMyAyNCAyNSAyNiAyNyAyOCAyOSAzMCAzMSAzMiAzMyAzNCAzNSAzNiAzNyAzOCAzOSA0MCA0MSA0MiA0MyA0NCA0NSA0NiA0NyA0OCA0OSA1MCA1MSA1MiA1MyA1NCA1NSA1NiA1NyA1OCA1OSA2MCA2MSA2MiA2MyA2NCA2NSA2NiA2NyA2OCA2OSA3MCA3MSA3MiA3MyA3NCA3NSA3NiA3NyA3OCA3OSA4MCA4MSA4MiA4MyA4NCA4NSA4NiA4NyA4OCA4OSA5MCA5MSA5MiA5MyA5NCA5NSA5NiA5NyA5OCA5OVxuIn1dLCIyIjpbeyJpbnB1dCI6IjEwXG4iLCJvdXRwdXQiOiI1NVxuIn0seyJpbnB1dCI6IjFcbiIsIm91dHB1dCI6IjFcbiJ9LHsiaW5wdXQiOiIyXG4iLCJvdXRwdXQiOiIzXG4ifSx7ImlucHV0IjoiM1xuIiwib3V0cHV0IjoiNlxuIn0seyJpbnB1dCI6IjEwXG4iLCJvdXRwdXQiOiI1NVxuIn0seyJpbnB1dCI6Ijk5XG4iLCJvdXRwdXQiOiI0OTUwXG4ifV0sIjMiOlt7ImlucHV0IjoiOFxuIiwib3V0cHV0IjoiMjBcbiJ9LHsiaW5wdXQiOiIxXG4iLCJvdXRwdXQiOiIwXG4ifSx7ImlucHV0IjoiMlxuIiwib3V0cHV0IjoiMlxuIn0seyJpbnB1dCI6IjNcbiIsIm91dHB1dCI6IjJcbiJ9LHsiaW5wdXQiOiIxMFxuIiwib3V0cHV0IjoiMzBcbiJ9LHsiaW5wdXQiOiI5OVxuIiwib3V0cHV0IjoiMjQ1MFxuIn1dLCI0IjpbeyJpbnB1dCI6IjVcbiIsIm91dHB1dCI6IjEyMFxuIn0seyJpbnB1dCI6IjBcbiIsIm91dHB1dCI6IjFcbiJ9LHsiaW5wdXQiOiIxXG4iLCJvdXRwdXQiOiIxXG4ifSx7ImlucHV0IjoiMlxuIiwib3V0cHV0IjoiMlxuIn0seyJpbnB1dCI6IjdcbiIsIm91dHB1dCI6IjUwNDBcbiJ9LHsiaW5wdXQiOiIxMlxuIiwib3V0cHV0IjoiNDc5MDAxNjAwXG4ifV0sIjUiOlt7ImlucHV0IjoiM1xuIiwib3V0cHV0IjoiMyA2IDkgMTIgMTUgMTggMjEgMjQgMjdcbiJ9LHsiaW5wdXQiOiIxXG4iLCJvdXRwdXQiOiIxIDIgMyA0IDUgNiA3IDggOVxuIn0seyJpbnB1dCI6IjJcbiIsIm91dHB1dCI6IjIgNCA2IDggMTAgMTIgMTQgMTYgMThcbiJ9LHsiaW5wdXQiOiI5XG4iLCJvdXRwdXQiOiI5IDE4IDI3IDM2IDQ1IDU0IDYzIDcyIDgxXG4ifSx7ImlucHV0IjoiMTBcbiIsIm91dHB1dCI6IjEwIDIwIDMwIDQwIDUwIDYwIDcwIDgwIDkwXG4ifSx7ImlucHV0IjoiOTlcbiIsIm91dHB1dCI6Ijk5IDE5OCAyOTcgMzk2IDQ5NSA1OTQgNjkzIDc5MiA4OTFcbiJ9XSwiNiI6W3siaW5wdXQiOiI0XG4iLCJvdXRwdXQiOiIzMFxuIn0seyJpbnB1dCI6IjFcbiIsIm91dHB1dCI6IjFcbiJ9LHsiaW5wdXQiOiIyXG4iLCJvdXRwdXQiOiI1XG4ifSx7ImlucHV0IjoiM1xuIiwib3V0cHV0IjoiMTRcbiJ9LHsiaW5wdXQiOiIxMFxuIiwib3V0cHV0IjoiMzg1XG4ifSx7ImlucHV0IjoiOTlcbiIsIm91dHB1dCI6IjMyODM1MFxuIn1dLCI3IjpbeyJpbnB1dCI6IjIwIDNcbiIsIm91dHB1dCI6IjZcbiJ9LHsiaW5wdXQiOiIxIDJcbiIsIm91dHB1dCI6IjBcbiJ9LHsiaW5wdXQiOiI5IDNcbiIsIm91dHB1dCI6IjNcbiJ9LHsiaW5wdXQiOiIyMCAyMVxuIiwib3V0cHV0IjoiMFxuIn0seyJpbnB1dCI6IjEwMCAxXG4iLCJvdXRwdXQiOiIxMDBcbiJ9LHsiaW5wdXQiOiIzNSA3XG4iLCJvdXRwdXQiOiI1XG4ifV0sIjgiOlt7ImlucHV0IjoiMiAzIDVcbiIsIm91dHB1dCI6IjIgNSA4IDExIDE0XG4ifSx7ImlucHV0IjoiMSAxIDFcbiIsIm91dHB1dCI6IjFcbiJ9LHsiaW5wdXQiOiIwIDAgNVxuIiwib3V0cHV0IjoiMCAwIDAgMCAwXG4ifSx7ImlucHV0IjoiLTUgMyA2XG4iLCJvdXRwdXQiOiItNSAtMiAxIDQgNyAxMFxuIn0seyJpbnB1dCI6IjEwIC0yIDVcbiIsIm91dHB1dCI6IjEwIDggNiA0IDJcbiJ9LHsiaW5wdXQiOiIxMDAgLTEwMCAzXG4iLCJvdXRwdXQiOiIxMDAgMCAtMTAwXG4ifV0sIjkiOlt7ImlucHV0IjoiMTJcbiIsIm91dHB1dCI6IjZcbiJ9LHsiaW5wdXQiOiIxXG4iLCJvdXRwdXQiOiIxXG4ifSx7ImlucHV0IjoiMlxuIiwib3V0cHV0IjoiMlxuIn0seyJpbnB1dCI6IjNcbiIsIm91dHB1dCI6IjJcbiJ9LHsiaW5wdXQiOiIxMFxuIiwib3V0cHV0IjoiNFxuIn0seyJpbnB1dCI6Ijk5XG4iLCJvdXRwdXQiOiI2XG4ifV0sIjEwIjpbeyJpbnB1dCI6IjE3XG4iLCJvdXRwdXQiOiJQUklNRVxuIn0seyJpbnB1dCI6IjBcbiIsIm91dHB1dCI6Ik5PVCBQUklNRVxuIn0seyJpbnB1dCI6IjFcbiIsIm91dHB1dCI6Ik5PVCBQUklNRVxuIn0seyJpbnB1dCI6IjJcbiIsIm91dHB1dCI6IlBSSU1FXG4ifSx7ImlucHV0IjoiNDlcbiIsIm91dHB1dCI6Ik5PVCBQUklNRVxuIn0seyJpbnB1dCI6Ijk5NzNcbiIsIm91dHB1dCI6IlBSSU1FXG4ifV0sIjExIjpbeyJpbnB1dCI6IjIgNVxuIiwib3V0cHV0IjoiMzJcbiJ9LHsiaW5wdXQiOiIwIDBcbiIsIm91dHB1dCI6IjFcbiJ9LHsiaW5wdXQiOiIwIDVcbiIsIm91dHB1dCI6IjBcbiJ9LHsiaW5wdXQiOiIxMCA4XG4iLCJvdXRwdXQiOiIxMDAwMDAwMDBcbiJ9LHsiaW5wdXQiOiIxIDhcbiIsIm91dHB1dCI6IjFcbiJ9LHsiaW5wdXQiOiIzIDRcbiIsIm91dHB1dCI6IjgxXG4ifV0sIjEyIjpbeyJpbnB1dCI6IjQwNzJcbiIsIm91dHB1dCI6IjEzXG4ifSx7ImlucHV0IjoiMFxuIiwib3V0cHV0IjoiMFxuIn0seyJpbnB1dCI6IjFcbiIsIm91dHB1dCI6IjFcbiJ9LHsiaW5wdXQiOiIxMFxuIiwib3V0cHV0IjoiMVxuIn0seyJpbnB1dCI6IjEwMDIwXG4iLCJvdXRwdXQiOiIzXG4ifSx7ImlucHV0IjoiOTk5OTk5XG4iLCJvdXRwdXQiOiI1NFxuIn1dLCIxMyI6W3siaW5wdXQiOiIxMjA0MFxuIiwib3V0cHV0IjoiNDAyMVxuIn0seyJpbnB1dCI6IjBcbiIsIm91dHB1dCI6IjBcbiJ9LHsiaW5wdXQiOiIxXG4iLCJvdXRwdXQiOiIxXG4ifSx7ImlucHV0IjoiMTBcbiIsIm91dHB1dCI6IjFcbiJ9LHsiaW5wdXQiOiIxMDAyMFxuIiwib3V0cHV0IjoiMjAwMVxuIn0seyJpbnB1dCI6Ijk5OTk5OVxuIiwib3V0cHV0IjoiOTk5OTk5XG4ifV0sIjE0IjpbeyJpbnB1dCI6IjRcbiIsIm91dHB1dCI6IiNcbiMjXG4jIyNcbiMjIyNcbiJ9LHsiaW5wdXQiOiIxXG4iLCJvdXRwdXQiOiIjXG4ifSx7ImlucHV0IjoiMlxuIiwib3V0cHV0IjoiI1xuIyNcbiJ9LHsiaW5wdXQiOiIzXG4iLCJvdXRwdXQiOiIjXG4jI1xuIyMjXG4ifSx7ImlucHV0IjoiNVxuIiwib3V0cHV0IjoiI1xuIyNcbiMjI1xuIyMjI1xuIyMjIyNcbiJ9LHsiaW5wdXQiOiI3XG4iLCJvdXRwdXQiOiIjXG4jI1xuIyMjXG4jIyMjXG4jIyMjI1xuIyMjIyMjXG4jIyMjIyMjXG4ifV0sIjE1IjpbeyJpbnB1dCI6IjNcbiIsIm91dHB1dCI6IjEgMiAzXG4yIDQgNlxuMyA2IDlcbiJ9LHsiaW5wdXQiOiIxXG4iLCJvdXRwdXQiOiIxXG4ifSx7ImlucHV0IjoiMlxuIiwib3V0cHV0IjoiMSAyXG4yIDRcbiJ9LHsiaW5wdXQiOiIzXG4iLCJvdXRwdXQiOiIxIDIgM1xuMiA0IDZcbjMgNiA5XG4ifSx7ImlucHV0IjoiNVxuIiwib3V0cHV0IjoiMSAyIDMgNCA1XG4yIDQgNiA4IDEwXG4zIDYgOSAxMiAxNVxuNCA4IDEyIDE2IDIwXG41IDEwIDE1IDIwIDI1XG4ifSx7ImlucHV0IjoiN1xuIiwib3V0cHV0IjoiMSAyIDMgNCA1IDYgN1xuMiA0IDYgOCAxMCAxMiAxNFxuMyA2IDkgMTIgMTUgMTggMjFcbjQgOCAxMiAxNiAyMCAyNCAyOFxuNSAxMCAxNSAyMCAyNSAzMCAzNVxuNiAxMiAxOCAyNCAzMCAzNiA0MlxuNyAxNCAyMSAyOCAzNSA0MiA0OVxuIn1dLCIxNiI6W3siaW5wdXQiOiI1XG4iLCJvdXRwdXQiOiI1IDQgMyAyIDEgMFxuIn0seyJpbnB1dCI6IjBcbiIsIm91dHB1dCI6IjBcbiJ9LHsiaW5wdXQiOiIxXG4iLCJvdXRwdXQiOiIxIDBcbiJ9LHsiaW5wdXQiOiIyXG4iLCJvdXRwdXQiOiIyIDEgMFxuIn0seyJpbnB1dCI6IjlcbiIsIm91dHB1dCI6IjkgOCA3IDYgNSA0IDMgMiAxIDBcbiJ9LHsiaW5wdXQiOiIzMFxuIiwib3V0cHV0IjoiMzAgMjkgMjggMjcgMjYgMjUgMjQgMjMgMjIgMjEgMjAgMTkgMTggMTcgMTYgMTUgMTQgMTMgMTIgMTEgMTAgOSA4IDcgNiA1IDQgMyAyIDEgMFxuIn1dLCIxNyI6W3siaW5wdXQiOiIzIC0yIDggMFxuIiwib3V0cHV0IjoiOVxuIn0seyJpbnB1dCI6IjBcbiIsIm91dHB1dCI6IjBcbiJ9LHsiaW5wdXQiOiIxIDBcbiIsIm91dHB1dCI6IjFcbiJ9LHsiaW5wdXQiOiItMyAzIDBcbiIsIm91dHB1dCI6IjBcbiJ9LHsiaW5wdXQiOiI4IC01IDEgMFxuIiwib3V0cHV0IjoiNFxuIn0seyJpbnB1dCI6IjEwMCAyMCAtMzAgNCAwXG4iLCJvdXRwdXQiOiI5NFxuIn1dLCIxOCI6W3siaW5wdXQiOiI1IDAgNyAtMVxuIiwib3V0cHV0IjoiM1xuIn0seyJpbnB1dCI6Ii0xXG4iLCJvdXRwdXQiOiIwXG4ifSx7ImlucHV0IjoiMCAtMVxuIiwib3V0cHV0IjoiMVxuIn0seyJpbnB1dCI6Ii00IDYgLTFcbiIsIm91dHB1dCI6IjJcbiJ9LHsiaW5wdXQiOiI0IDMgMiAxIDAgLTFcbiIsIm91dHB1dCI6IjVcbiJ9LHsiaW5wdXQiOiI3IDcgNyA3IC0xXG4iLCJvdXRwdXQiOiI0XG4ifV0sIjE5IjpbeyJpbnB1dCI6IjIgOSA0IC0xXG4iLCJvdXRwdXQiOiI5XG4ifSx7ImlucHV0IjoiLTFcbiIsIm91dHB1dCI6IkVNUFRZXG4ifSx7ImlucHV0IjoiMCAtMVxuIiwib3V0cHV0IjoiMFxuIn0seyJpbnB1dCI6IjQgNCAtMVxuIiwib3V0cHV0IjoiNFxuIn0seyJpbnB1dCI6IjAgNyAwIDYgLTFcbiIsIm91dHB1dCI6IjdcbiJ9LHsiaW5wdXQiOiIxMDAwMCAzIDkgLTFcbiIsIm91dHB1dCI6IjEwMDAwXG4ifV0sIjIwIjpbeyJpbnB1dCI6IjI0IDM2XG4iLCJvdXRwdXQiOiIxMlxuIn0seyJpbnB1dCI6IjEgMVxuIiwib3V0cHV0IjoiMVxuIn0seyJpbnB1dCI6IjMgNVxuIiwib3V0cHV0IjoiMVxuIn0seyJpbnB1dCI6IjI1IDEwXG4iLCJvdXRwdXQiOiI1XG4ifSx7ImlucHV0IjoiOTYgMzZcbiIsIm91dHB1dCI6IjEyXG4ifSx7ImlucHV0IjoiOTk5OTEgOTk5ODlcbiIsIm91dHB1dCI6IjFcbiJ9XSwiMjEiOlt7ImlucHV0IjoiNlxuIiwib3V0cHV0IjoiOFxuIn0seyJpbnB1dCI6IjFcbiIsIm91dHB1dCI6IjBcbiJ9LHsiaW5wdXQiOiIyXG4iLCJvdXRwdXQiOiIxXG4ifSx7ImlucHV0IjoiM1xuIiwib3V0cHV0IjoiN1xuIn0seyJpbnB1dCI6IjI3XG4iLCJvdXRwdXQiOiIxMTFcbiJ9LHsiaW5wdXQiOiIxMDAwMFxuIiwib3V0cHV0IjoiMjlcbiJ9XSwiMjIiOlt7ImlucHV0IjoiNzA4MDRcbiIsIm91dHB1dCI6IjVcbiJ9LHsiaW5wdXQiOiIwXG4iLCJvdXRwdXQiOiIxXG4ifSx7ImlucHV0IjoiMVxuIiwib3V0cHV0IjoiMVxuIn0seyJpbnB1dCI6IjEwXG4iLCJvdXRwdXQiOiIyXG4ifSx7ImlucHV0IjoiMTAwMjBcbiIsIm91dHB1dCI6IjVcbiJ9LHsiaW5wdXQiOiI5OTk5OTlcbiIsIm91dHB1dCI6IjZcbiJ9XSwiMjMiOlt7ImlucHV0IjoiMTNcbiIsIm91dHB1dCI6IjExMDFcbiJ9LHsiaW5wdXQiOiIwXG4iLCJvdXRwdXQiOiIwXG4ifSx7ImlucHV0IjoiMVxuIiwib3V0cHV0IjoiMVxuIn0seyJpbnB1dCI6IjEwXG4iLCJvdXRwdXQiOiIxMDEwXG4ifSx7ImlucHV0IjoiMTAwMjBcbiIsIm91dHB1dCI6IjEwMDExMTAwMTAwMTAwXG4ifSx7ImlucHV0IjoiOTk5OTk5XG4iLCJvdXRwdXQiOiIxMTExMDEwMDAwMTAwMDExMTExMVxuIn1dLCIyNCI6W3siaW5wdXQiOiI5ODc1XG4iLCJvdXRwdXQiOiIyXG4ifSx7ImlucHV0IjoiMFxuIiwib3V0cHV0IjoiMFxuIn0seyJpbnB1dCI6IjFcbiIsIm91dHB1dCI6IjFcbiJ9LHsiaW5wdXQiOiIxMFxuIiwib3V0cHV0IjoiMVxuIn0seyJpbnB1dCI6IjEwMDIwXG4iLCJvdXRwdXQiOiIzXG4ifSx7ImlucHV0IjoiOTk5OTk5XG4iLCJvdXRwdXQiOiI5XG4ifV0sIjI1IjpbeyJpbnB1dCI6IjEwIDcgMzVcbiIsIm91dHB1dCI6IjRcbiJ9LHsiaW5wdXQiOiIwIDEgMFxuIiwib3V0cHV0IjoiMFxuIn0seyJpbnB1dCI6IjEwIDEgMTBcbiIsIm91dHB1dCI6IjBcbiJ9LHsiaW5wdXQiOiIwIDMgMTBcbiIsIm91dHB1dCI6IjRcbiJ9LHsiaW5wdXQiOiIyMCA0IDIxXG4iLCJvdXRwdXQiOiIxXG4ifSx7ImlucHV0IjoiMiA3IDUwXG4iLCJvdXRwdXQiOiI3XG4ifV0sIjI2IjpbeyJpbnB1dCI6IjVcbjEgMiAzIDQgNVxuIiwib3V0cHV0IjoiMTVcbiJ9LHsiaW5wdXQiOiIxXG43XG4iLCJvdXRwdXQiOiI3XG4ifSx7ImlucHV0IjoiNFxuLTMgLTMgLTMgLTNcbiIsIm91dHB1dCI6Ii0xMlxuIn0seyJpbnB1dCI6IjVcbjAgMCAxIC0xIDBcbiIsIm91dHB1dCI6IjBcbiJ9LHsiaW5wdXQiOiI2XG4xIDIgMyA0IDUgNlxuIiwib3V0cHV0IjoiMjFcbiJ9LHsiaW5wdXQiOiIxNVxuMTIgMTIgLTMgLTQgNyAtMTIgMiAtMTIgNSAtMSA4IDQgLTQgLTQgOFxuIiwib3V0cHV0IjoiMThcbiJ9XSwiMjciOlt7ImlucHV0IjoiNVxuLTMgOCAyIDggLTVcbiIsIm91dHB1dCI6IjggLTVcbiJ9LHsiaW5wdXQiOiIxXG43XG4iLCJvdXRwdXQiOiI3IDdcbiJ9LHsiaW5wdXQiOiI0XG4tMyAtMyAtMyAtM1xuIiwib3V0cHV0IjoiLTMgLTNcbiJ9LHsiaW5wdXQiOiI1XG4wIDAgMSAtMSAwXG4iLCJvdXRwdXQiOiIxIC0xXG4ifSx7ImlucHV0IjoiNlxuMSAyIDMgNCA1IDZcbiIsIm91dHB1dCI6IjYgMVxuIn0seyJpbnB1dCI6IjE1XG4tNyAtNCAtNiAtNSAwIDYgMTEgLTkgLTEwIC0yIDEgLTUgLTExIDggLTExXG4iLCJvdXRwdXQiOiIxMSAtMTFcbiJ9XSwiMjgiOlt7ImlucHV0IjoiNlxuLTIgMCAzIDUgMCAtMVxuIiwib3V0cHV0IjoiMiAyIDJcbiJ9LHsiaW5wdXQiOiIxXG43XG4iLCJvdXRwdXQiOiIxIDAgMFxuIn0seyJpbnB1dCI6IjRcbi0zIC0zIC0zIC0zXG4iLCJvdXRwdXQiOiIwIDQgMFxuIn0seyJpbnB1dCI6IjVcbjAgMCAxIC0xIDBcbiIsIm91dHB1dCI6IjEgMSAzXG4ifSx7ImlucHV0IjoiNlxuMSAyIDMgNCA1IDZcbiIsIm91dHB1dCI6IjYgMCAwXG4ifSx7ImlucHV0IjoiMTVcbi0xIC0xIC04IDUgMiAtOSA0IDggMCA3IDExIDUgOSAxMiAtNVxuIiwib3V0cHV0IjoiOSA1IDFcbiJ9XSwiMjkiOlt7ImlucHV0IjoiNVxuNCAyIDcgMiA4XG4yXG4iLCJvdXRwdXQiOiIxXG4ifSx7ImlucHV0IjoiMVxuN1xuN1xuIiwib3V0cHV0IjoiMFxuIn0seyJpbnB1dCI6IjRcbi0zIC0zIC0zIC0zXG4tM1xuIiwib3V0cHV0IjoiMFxuIn0seyJpbnB1dCI6IjVcbjAgMCAxIC0xIDBcbjk5XG4iLCJvdXRwdXQiOiItMVxuIn0seyJpbnB1dCI6IjZcbjEgMiAzIDQgNSA2XG4xXG4iLCJvdXRwdXQiOiIwXG4ifSx7ImlucHV0IjoiMTVcbjIgOCAxMiAtMiA5IC0xMCAtNSA4IDEyIDEwIC04IC05IC03IC0xMCAtMTJcbi0xMlxuIiwib3V0cHV0IjoiMTRcbiJ9XSwiMzAiOlt7ImlucHV0IjoiNlxuMSA0IDEgMiAxIDNcbjFcbiIsIm91dHB1dCI6IjRcbiJ9LHsiaW5wdXQiOiIxXG43XG43XG4iLCJvdXRwdXQiOiIwXG4ifSx7ImlucHV0IjoiNFxuLTMgLTMgLTMgLTNcbi0zXG4iLCJvdXRwdXQiOiIzXG4ifSx7ImlucHV0IjoiNVxuMCAwIDEgLTEgMFxuOTlcbiIsIm91dHB1dCI6Ii0xXG4ifSx7ImlucHV0IjoiNlxuMSAyIDMgNCA1IDZcbjFcbiIsIm91dHB1dCI6IjBcbiJ9LHsiaW5wdXQiOiIxNVxuMTEgLTUgMTIgNSAxMCAtMSAtNCAtNiAtMTEgLTMgLTcgLTYgNiAtNiAxMVxuMTFcbiIsIm91dHB1dCI6IjE0XG4ifV0sIjMxIjpbeyJpbnB1dCI6IjVcbjEgMiAzIDQgNVxuIiwib3V0cHV0IjoiNSA0IDMgMiAxXG4ifSx7ImlucHV0IjoiMVxuN1xuIiwib3V0cHV0IjoiN1xuIn0seyJpbnB1dCI6IjRcbi0zIC0zIC0zIC0zXG4iLCJvdXRwdXQiOiItMyAtMyAtMyAtM1xuIn0seyJpbnB1dCI6IjVcbjAgMCAxIC0xIDBcbiIsIm91dHB1dCI6IjAgLTEgMSAwIDBcbiJ9LHsiaW5wdXQiOiI2XG4xIDIgMyA0IDUgNlxuIiwib3V0cHV0IjoiNiA1IDQgMyAyIDFcbiJ9LHsiaW5wdXQiOiIxNVxuLTExIDEwIC01IC0yIC05IDAgNSAtOCA5IDQgLTIgLTYgMCAyIDFcbiIsIm91dHB1dCI6IjEgMiAwIC02IC0yIDQgOSAtOCA1IDAgLTkgLTIgLTUgMTAgLTExXG4ifV0sIjMyIjpbeyJpbnB1dCI6IjVcbjIgNSAtNCAwIDNcbiIsIm91dHB1dCI6IjMgMlxuIn0seyJpbnB1dCI6IjFcbjdcbiIsIm91dHB1dCI6IjAgMVxuIn0seyJpbnB1dCI6IjRcbi0zIC0zIC0zIC0zXG4iLCJvdXRwdXQiOiIwIDRcbiJ9LHsiaW5wdXQiOiI1XG4wIDAgMSAtMSAwXG4iLCJvdXRwdXQiOiIzIDJcbiJ9LHsiaW5wdXQiOiI2XG4xIDIgMyA0IDUgNlxuIiwib3V0cHV0IjoiMyAzXG4ifSx7ImlucHV0IjoiMTVcbjEgMyAtMTAgMCAtMTEgMCAtOCAtNCA3IC03IDUgLTQgMTAgLTEwIC0xMVxuIiwib3V0cHV0IjoiOCA3XG4ifV0sIjMzIjpbeyJpbnB1dCI6IjRcbjEgMiAzIDRcbiIsIm91dHB1dCI6IjIuNTBcbiJ9LHsiaW5wdXQiOiIxXG4wXG4iLCJvdXRwdXQiOiIwLjAwXG4ifSx7ImlucHV0IjoiM1xuMyAzIDNcbiIsIm91dHB1dCI6IjMuMDBcbiJ9LHsiaW5wdXQiOiIzXG4wIDAgMVxuIiwib3V0cHV0IjoiMC4zM1xuIn0seyJpbnB1dCI6IjRcbjEgMiAzIDRcbiIsIm91dHB1dCI6IjIuNTBcbiJ9LHsiaW5wdXQiOiI1XG4yNTAgNzUwIDEwMDAgMCAwXG4iLCJvdXRwdXQiOiI0MDAuMDBcbiJ9XSwiMzQiOlt7ImlucHV0IjoiNVxuMiA0IDYgOCAxMFxuIiwib3V0cHV0IjoiMlxuIn0seyJpbnB1dCI6IjFcbjdcbiIsIm91dHB1dCI6IjBcbiJ9LHsiaW5wdXQiOiI0XG4tMyAtMyAtMyAtM1xuIiwib3V0cHV0IjoiMFxuIn0seyJpbnB1dCI6IjVcbjAgMCAxIC0xIDBcbiIsIm91dHB1dCI6IjFcbiJ9LHsiaW5wdXQiOiI2XG4xIDIgMyA0IDUgNlxuIiwib3V0cHV0IjoiM1xuIn0seyJpbnB1dCI6IjE1XG4tNCA0IC03IDggMCAtNSAxMCAtNiAxIC02IDEgOSAtNCA0IDEyXG4iLCJvdXRwdXQiOiI2XG4ifV0sIjM1IjpbeyJpbnB1dCI6IjZcbjkgMiA5IDUgNSAzXG4iLCJvdXRwdXQiOiI1XG4ifSx7ImlucHV0IjoiMVxuLTVcbiIsIm91dHB1dCI6Ik5PTkVcbiJ9LHsiaW5wdXQiOiIzXG4zIDMgM1xuIiwib3V0cHV0IjoiTk9ORVxuIn0seyJpbnB1dCI6IjVcbjIgNSA1IDIgM1xuIiwib3V0cHV0IjoiM1xuIn0seyJpbnB1dCI6IjNcbi03IC0yIC00XG4iLCJvdXRwdXQiOiItNFxuIn0seyJpbnB1dCI6IjNcbi0xMDAwIDEwMDAgLTk5OVxuIiwib3V0cHV0IjoiLTk5OVxuIn1dLCIzNiI6W3siaW5wdXQiOiI1XG4xIDIgMiA0IDlcbiIsIm91dHB1dCI6IllFU1xuIn0seyJpbnB1dCI6IjFcbjFcbiIsIm91dHB1dCI6IllFU1xuIn0seyJpbnB1dCI6IjNcbjEgMSAxXG4iLCJvdXRwdXQiOiJZRVNcbiJ9LHsiaW5wdXQiOiIzXG4xIDIgM1xuIiwib3V0cHV0IjoiWUVTXG4ifSx7ImlucHV0IjoiM1xuMyAyIDFcbiIsIm91dHB1dCI6Ik5PXG4ifSx7ImlucHV0IjoiNVxuLTMgLTMgMCAzIDRcbiIsIm91dHB1dCI6IllFU1xuIn1dLCIzNyI6W3siaW5wdXQiOiI2XG4zIDEgMyAzIDIgNFxuM1xuIiwib3V0cHV0IjoiM1xuIn0seyJpbnB1dCI6IjFcbjdcbjdcbiIsIm91dHB1dCI6IjFcbiJ9LHsiaW5wdXQiOiI0XG4tMyAtMyAtMyAtM1xuLTNcbiIsIm91dHB1dCI6IjRcbiJ9LHsiaW5wdXQiOiI1XG4wIDAgMSAtMSAwXG45OVxuIiwib3V0cHV0IjoiMFxuIn0seyJpbnB1dCI6IjZcbjEgMiAzIDQgNSA2XG4xXG4iLCJvdXRwdXQiOiIxXG4ifSx7ImlucHV0IjoiMTVcbi04IC0xMCAtOCA1IC00IC0yIC00IC03IC00IDggMTEgNyA1IC0xMSA3XG43XG4iLCJvdXRwdXQiOiIyXG4ifV0sIjM4IjpbeyJpbnB1dCI6IjZcbjEgMSAyIDIgMiAzXG4iLCJvdXRwdXQiOiIzXG4ifSx7ImlucHV0IjoiMVxuNVxuIiwib3V0cHV0IjoiMFxuIn0seyJpbnB1dCI6IjJcbjIgMlxuIiwib3V0cHV0IjoiMVxuIn0seyJpbnB1dCI6IjNcbjAgMCAwXG4iLCJvdXRwdXQiOiIyXG4ifSx7ImlucHV0IjoiNFxuMSAyIDMgNFxuIiwib3V0cHV0IjoiMFxuIn0seyJpbnB1dCI6IjZcbjEgMSAyIDMgMyAzXG4iLCJvdXRwdXQiOiIzXG4ifV0sIjM5IjpbeyJpbnB1dCI6IjRcbjIgMyAxIDRcbiIsIm91dHB1dCI6IjIgNSA2IDEwXG4ifSx7ImlucHV0IjoiMVxuN1xuIiwib3V0cHV0IjoiN1xuIn0seyJpbnB1dCI6IjRcbi0zIC0zIC0zIC0zXG4iLCJvdXRwdXQiOiItMyAtNiAtOSAtMTJcbiJ9LHsiaW5wdXQiOiI1XG4wIDAgMSAtMSAwXG4iLCJvdXRwdXQiOiIwIDAgMSAwIDBcbiJ9LHsiaW5wdXQiOiI2XG4xIDIgMyA0IDUgNlxuIiwib3V0cHV0IjoiMSAzIDYgMTAgMTUgMjFcbiJ9LHsiaW5wdXQiOiIxNVxuMTAgLTkgLTkgLTYgLTIgLTUgLTggMSAtNCAtMTIgLTUgMSAtOCAtNyAtOVxuIiwib3V0cHV0IjoiMTAgMSAtOCAtMTQgLTE2IC0yMSAtMjkgLTI4IC0zMiAtNDQgLTQ5IC00OCAtNTYgLTYzIC03MlxuIn1dLCI0MCI6W3siaW5wdXQiOiI1XG4xIDIgMyA0IDVcbiIsIm91dHB1dCI6IjIgMyA0IDUgMVxuIn0seyJpbnB1dCI6IjFcbjdcbiIsIm91dHB1dCI6IjdcbiJ9LHsiaW5wdXQiOiI0XG4tMyAtMyAtMyAtM1xuIiwib3V0cHV0IjoiLTMgLTMgLTMgLTNcbiJ9LHsiaW5wdXQiOiI1XG4wIDAgMSAtMSAwXG4iLCJvdXRwdXQiOiIwIDEgLTEgMCAwXG4ifSx7ImlucHV0IjoiNlxuMSAyIDMgNCA1IDZcbiIsIm91dHB1dCI6IjIgMyA0IDUgNiAxXG4ifSx7ImlucHV0IjoiMTVcbi0xIDkgMiA5IDUgLTggLTEwIC00IC04IDMgMyAzIDEwIC0yIDdcbiIsIm91dHB1dCI6IjkgMiA5IDUgLTggLTEwIC00IC04IDMgMyAzIDEwIC0yIDcgLTFcbiJ9XSwiNDEiOlt7ImlucHV0IjoiNSAyXG4xIDIgMyA0IDVcbiIsIm91dHB1dCI6IjQgNSAxIDIgM1xuIn0seyJpbnB1dCI6IjEgMFxuOFxuIiwib3V0cHV0IjoiOFxuIn0seyJpbnB1dCI6IjIgMVxuMSAyXG4iLCJvdXRwdXQiOiIyIDFcbiJ9LHsiaW5wdXQiOiI0IDRcbjEgMiAzIDRcbiIsIm91dHB1dCI6IjEgMiAzIDRcbiJ9LHsiaW5wdXQiOiI1IDEwMlxuNSA1IDEgMCAyXG4iLCJvdXRwdXQiOiIwIDIgNSA1IDFcbiJ9LHsiaW5wdXQiOiI2IDEwMDAwXG40IDggMTUgMTYgMjMgNDJcbiIsIm91dHB1dCI6IjE1IDE2IDIzIDQyIDQgOFxuIn1dLCI0MiI6W3siaW5wdXQiOiI3XG4yIDMgMiA0IDMgNSA0XG4iLCJvdXRwdXQiOiIyIDMgNCA1XG4ifSx7ImlucHV0IjoiMVxuN1xuIiwib3V0cHV0IjoiN1xuIn0seyJpbnB1dCI6IjRcbi0zIC0zIC0zIC0zXG4iLCJvdXRwdXQiOiItM1xuIn0seyJpbnB1dCI6IjVcbjAgMCAxIC0xIDBcbiIsIm91dHB1dCI6IjAgMSAtMVxuIn0seyJpbnB1dCI6IjZcbjEgMiAzIDQgNSA2XG4iLCJvdXRwdXQiOiIxIDIgMyA0IDUgNlxuIn0seyJpbnB1dCI6IjE1XG4tNyA5IDMgLTEyIDQgLTYgLTEwIC01IDggNCAyIC05IDggLTkgLTEwXG4iLCJvdXRwdXQiOiItNyA5IDMgLTEyIDQgLTYgLTEwIC01IDggMiAtOVxuIn1dLCI0MyI6W3siaW5wdXQiOiI2XG4xIDEgMiAzIDIgNFxuIiwib3V0cHV0IjoiNFxuIn0seyJpbnB1dCI6IjFcbjdcbiIsIm91dHB1dCI6IjFcbiJ9LHsiaW5wdXQiOiI0XG4tMyAtMyAtMyAtM1xuIiwib3V0cHV0IjoiMVxuIn0seyJpbnB1dCI6IjVcbjAgMCAxIC0xIDBcbiIsIm91dHB1dCI6IjNcbiJ9LHsiaW5wdXQiOiI2XG4xIDIgMyA0IDUgNlxuIiwib3V0cHV0IjoiNlxuIn0seyJpbnB1dCI6IjE1XG43IDAgLTExIDggLTMgLTMgOSAtMTEgMiAtMTAgLTkgOSA2IDUgMlxuIiwib3V0cHV0IjoiMTFcbiJ9XSwiNDQiOlt7ImlucHV0IjoiN1xuMSAzIDUgMiA0IDYgOFxuIiwib3V0cHV0IjoiNFxuIn0seyJpbnB1dCI6IjFcbjVcbiIsIm91dHB1dCI6IjFcbiJ9LHsiaW5wdXQiOiIzXG4xIDEgMVxuIiwib3V0cHV0IjoiMVxuIn0seyJpbnB1dCI6IjRcbjEgMiAzIDRcbiIsIm91dHB1dCI6IjRcbiJ9LHsiaW5wdXQiOiI0XG40IDMgMiAxXG4iLCJvdXRwdXQiOiIxXG4ifSx7ImlucHV0IjoiN1xuMSAzIDIgNCA2IDAgMVxuIiwib3V0cHV0IjoiM1xuIn1dLCI0NSI6W3siaW5wdXQiOiI5XG4yIDQgMSA2IDggMTAgMyAyIDRcbiIsIm91dHB1dCI6IjNcbiJ9LHsiaW5wdXQiOiIxXG4xXG4iLCJvdXRwdXQiOiIwXG4ifSx7ImlucHV0IjoiMVxuMlxuIiwib3V0cHV0IjoiMVxuIn0seyJpbnB1dCI6IjRcbjEgMyA1IDdcbiIsIm91dHB1dCI6IjBcbiJ9LHsiaW5wdXQiOiI0XG4yIDQgNiA4XG4iLCJvdXRwdXQiOiI0XG4ifSx7ImlucHV0IjoiOFxuMSAyIDMgNCA2IDggMSAyXG4iLCJvdXRwdXQiOiIzXG4ifV0sIjQ2IjpbeyJpbnB1dCI6IjUgNlxuMSA1IDIgNCAzXG4iLCJvdXRwdXQiOiIyXG4ifSx7ImlucHV0IjoiMSAwXG4wXG4iLCJvdXRwdXQiOiIwXG4ifSx7ImlucHV0IjoiMiA0XG4yIDJcbiIsIm91dHB1dCI6IjFcbiJ9LHsiaW5wdXQiOiIzIDBcbjAgMCAwXG4iLCJvdXRwdXQiOiIzXG4ifSx7ImlucHV0IjoiNSA1XG4xIDIgMyA0IDVcbiIsIm91dHB1dCI6IjJcbiJ9LHsiaW5wdXQiOiI2IDBcbjMgLTMgNSAtNSAwIDBcbiIsIm91dHB1dCI6IjNcbiJ9XSwiNDciOlt7ImlucHV0IjoiNVxuMSAyIDQgNVxuIiwib3V0cHV0IjoiM1xuIn0seyJpbnB1dCI6IjJcbjJcbiIsIm91dHB1dCI6IjFcbiJ9LHsiaW5wdXQiOiIzXG4yIDFcbiIsIm91dHB1dCI6IjNcbiJ9LHsiaW5wdXQiOiI1XG4zIDUgNCAxXG4iLCJvdXRwdXQiOiIyXG4ifSx7ImlucHV0IjoiMjBcbjYgMTEgMTkgMTIgMyAxNCA4IDE2IDE4IDE3IDcgNSAxNSA5IDIwIDIgMTAgMSA0XG4iLCJvdXRwdXQiOiIxM1xuIn0seyJpbnB1dCI6IjEwMFxuMjUgMzMgMjEgNzcgNjggNTEgODIgOTIgMzEgMzAgNzEgMzcgOTQgMTAgNjMgODYgODEgNDQgMjcgMiA0NiAxNCA5MCAyOSA2MSAyNCA2OSAxNyA2MCA4MyA0MSA4NSA3NSA1NCAzNCA3NiA4MCA4NCA1NyA5NyA5IDQ4IDY1IDM1IDMgNSA3OSAxOSA2NyAxMyA5NiA5MSAyMyAyMiA0IDU2IDk4IDMyIDM5IDY0IDggNzMgMjAgODcgOTkgNDcgNzQgNiA1NSA3MCA0MiAyOCAyNiA5NSA0OSA3MiA0NSA2NiA5MyA1OCA1MiAzOCA2MiA1OSA4OCAxMiA1MCAxMSAxIDM2IDc4IDUzIDg5IDQwIDE4IDcgMTYgNDMgMTVcbiIsIm91dHB1dCI6IjEwMFxuIn1dLCI0OCI6W3siaW5wdXQiOiI1XG4zIDcgOSA3IDNcbiIsIm91dHB1dCI6IllFU1xuIn0seyJpbnB1dCI6IjFcbjFcbiIsIm91dHB1dCI6IllFU1xuIn0seyJpbnB1dCI6IjJcbjEgMVxuIiwib3V0cHV0IjoiWUVTXG4ifSx7ImlucHV0IjoiM1xuMSAyIDFcbiIsIm91dHB1dCI6IllFU1xuIn0seyJpbnB1dCI6IjRcbjEgMiAzIDRcbiIsIm91dHB1dCI6Ik5PXG4ifSx7ImlucHV0IjoiNFxuMCAtMSAtMSAwXG4iLCJvdXRwdXQiOiJZRVNcbiJ9XSwiNDkiOlt7ImlucHV0IjoiNVxuMSAzIDUgMiAyXG4iLCJvdXRwdXQiOiIyXG4ifSx7ImlucHV0IjoiMVxuMFxuIiwib3V0cHV0IjoiMFxuIn0seyJpbnB1dCI6IjJcbjEgMVxuIiwib3V0cHV0IjoiLTFcbiJ9LHsiaW5wdXQiOiIzXG4xIDAgMVxuIiwib3V0cHV0IjoiMVxuIn0seyJpbnB1dCI6IjNcbjEgMiAzXG4iLCJvdXRwdXQiOiItMVxuIn0seyJpbnB1dCI6IjRcbjEgMiAzIDNcbiIsIm91dHB1dCI6IjJcbiJ9XSwiNTAiOlt7ImlucHV0IjoiNlxuNSAxIDQgMiA2IDNcbiIsIm91dHB1dCI6IjEgMiAzIDQgNSA2XG4ifSx7ImlucHV0IjoiMVxuN1xuIiwib3V0cHV0IjoiN1xuIn0seyJpbnB1dCI6IjRcbi0zIC0zIC0zIC0zXG4iLCJvdXRwdXQiOiItMyAtMyAtMyAtM1xuIn0seyJpbnB1dCI6IjVcbjAgMCAxIC0xIDBcbiIsIm91dHB1dCI6Ii0xIDAgMCAwIDFcbiJ9LHsiaW5wdXQiOiI2XG4xIDIgMyA0IDUgNlxuIiwib3V0cHV0IjoiMSAyIDMgNCA1IDZcbiJ9LHsiaW5wdXQiOiIxNVxuLTIgNSAzIDIgLTYgMCAwIDExIDcgLTEwIDEgLTEgMCAtNiAyXG4iLCJvdXRwdXQiOiItMTAgLTYgLTYgLTIgLTEgMCAwIDAgMSAyIDIgMyA1IDcgMTFcbiJ9XX0=').decode('utf-8'))
_STATUS={}

def _check_one(q, verbose=True):
    q=int(str(q).upper().lstrip('Q'))
    if not 1<=q<=50:raise ValueError('題號必須介於 1 和 50')
    path=_ANSWER_DIR/f'q{q:02d}.c'
    if not path.exists():
        _STATUS[q]='未作答'
        if verbose:print(f'Q{q:02d} 未作答：請先執行 %writefile 程式格')
        return False
    if not _COMPILER:
        _STATUS[q]='環境未完成'
        if verbose:print(f'Q{q:02d} 無編譯器：請先安裝 cc/gcc/clang')
        return False
    src=path.read_text(encoding='utf-8')
    if '/* TODO:' in src:
        _STATUS[q]='未作答'
        if verbose:print(f'Q{q:02d} 未作答：請完成程式，刪除 TODO 註解並重新執行 %writefile')
        return False
    with tempfile.TemporaryDirectory() as tmp:
        exe=Path(tmp)/'student_program'
        cmd=[_COMPILER,'-std=c11','-O0','-Wall','-Wextra',str(path),'-o',str(exe)]
        try:
            compiled=subprocess.run(cmd,text=True,capture_output=True,timeout=12)
        except subprocess.TimeoutExpired:
            _STATUS[q]='CE'
            if verbose:print(f'Q{q:02d} 編譯逾時')
            return False
        if compiled.returncode:
            _STATUS[q]='CE'
            if verbose:print(f'Q{q:02d} 編譯失敗 (CE):\n'+compiled.stderr[-2500:])
            return False
        cases=_TEST_CASES[str(q)]
        for i,test in enumerate(cases):
            try:
                res=subprocess.run([str(exe)],input=test['input'],text=True,capture_output=True,timeout=2)
            except subprocess.TimeoutExpired:
                _STATUS[q]='TLE'
                if verbose:print(f'Q{q:02d} 執行逾時 (TLE)，第 {i+1} 組測資')
                return False
            if res.returncode:
                _STATUS[q]='RE'
                if verbose:print(f'Q{q:02d} 執行錯誤 (RE)，第 {i+1} 組，return code={res.returncode}；'+res.stderr[-700:])
                return False
            if res.stdout.split()!=test['output'].split():
                _STATUS[q]='WA'
                if verbose:
                    print(f'Q{q:02d} 答案錯誤 (WA)：'+('公開範例不符合' if i==0 else '第 '+str(i+1)+' 組額外測資不符合'))
                    if i==0:
                        print('輸入:\n'+test['input'])
                        print('預期輸出:\n'+test['output'])
                        print('實際輸出:\n'+res.stdout)
                    else:print('請檢查邊界情況、迴圈條件、陣列索引及輸出格式。')
                return False
    _STATUS[q]='AC'
    if verbose:print(f'Q{q:02d} ✅ AC：公開範例 + {len(cases)-1} 組額外測資全部通過')
    return True

def run_public(q):
    return _check_one(q,verbose=True)

def run_all_checks():
    print('=== C 語言 50 題總檢查 ===')
    total=0
    for q in range(1,51):
        ok=_check_one(q,verbose=False)
        total+=int(ok)
        print(f'Q{q:02d} {_STATUS[q]}',end='    ')
        if q%5==0:print()
    print(f'\n總通過：{total}/50；未通過：{50-total}/50')
    with open(_ANSWER_DIR/'progress.csv','w',encoding='utf-8-sig',newline='') as f:
        writer=csv.writer(f)
        writer.writerow(['題號','結果'])
        writer.writerows([(f'Q{q:02d}',_STATUS.get(q,'未檢查')) for q in range(1,51)])
    print('進度紀錄：',_ANSWER_DIR/'progress.csv')
    return total
````
